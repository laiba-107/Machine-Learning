```python
from keras import Sequential 
from keras.layers import Dense, SimpleRNN
```

```python
model = Sequential()
model.add(SimpleRNN(units=3, input_shape=(4, 5)))
model.add(Dense(1, activation="sigmoid"))
```

```python
model.summary()
```

```python
rnn_weights = model.layers[0].get_weights()

Wx = rnn_weights[0]   # Input weights
Wh = rnn_weights[1]   # Recurrent weights
b  = rnn_weights[2]   # Bias

print("Wx shape:", Wx.shape)
print("Wh shape:", Wh.shape)
print("b shape:", b.shape)
```
