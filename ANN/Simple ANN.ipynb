```python
import math
```

### Activation Functions

```python
def sigmoid(x):
    return 1/ (1 + math.exp(-x))
```

```python
def softmax(z_list):
    exps = [math.exp(z) for z in z_list]
    sum_exps = sum(exps)
    return [e/sum_exps for e in exps]
```

### Inputs

```python
X = [[1,4],[0,0.7],[6,0.9],[0.8,2],[3,6]]
Y = [1,0,0,1,2]
X
```

### Wegihts

```python
# Hidden layer weights & biases
w1, w2, b1 = 0.2, 0.3, 0.1
w3, w4, b2 = 0.4, 0.1, 0.2
```

```python
# Output layer weights (2 hidden → 3 outputs)
w5, w6, b3 = 0.5, 0.6, 0.05    # class 0
w7, w8, b4 = 0.1, 0.3, -0.1    # class 1
w9, w10, b5 = 0.2, 0.4, 0.2    # class 2
```

### To Store Loss

```python
losses = []
```

```python
for i in range(len(X)):
    x1, x2 = X[i]
    
    # Hidden layer
    h1 = sigmoid(x1*w1 + x2*w2 + b1)
    h2 = sigmoid(x1*w3 + x2*w4 + b2)
    
    # Output layer (3 neurons for 3 classes)
    z_out = [
        h1*w5 + h2*w6 + b3,  # class 0
        h1*w7 + h2*w8 + b4,  # class 1
        h1*w9 + h2*w10 + b5  # class 2
    ]
    
    y_hat = softmax(z_out)
    
    # One-hot encode target
    y_true = [0,0,0]
    y_true[Y[i]] = 1
    
    # Loss (Categorical Cross-Entropy)
    loss = -sum(y_true[j]*math.log(y_hat[j]+1e-9) for j in range(3))  # +1e-9 to avoid log(0)
    losses.append(loss)
    
    print(f"Sample {i+1}: predicted={[round(v,3) for v in y_hat]}, target={Y[i]}, loss={round(loss,3)}")
```

### Cost function (mean loss)

```python
cost = sum(losses)/len(losses)
print(f"\nCost (Categorical Cross-Entropy over all samples): {round(cost,3)}")
```
