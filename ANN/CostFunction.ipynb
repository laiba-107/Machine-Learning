# Cost Functions For Regression

### 1. Mean Squared Error (MSE): 
********

```python
import numpy as np
```

### Actual VS Predicted

```python
y_pred = np.array([1.2, 2.1, 3.5, 4.8])
y_actual = np.array([1, 2, 3, 5])
```

```python
n = len(y_pred)
```

```python
mse = 1/n * np.sum((y_pred - y_actual)**2)
print('MSE:', mse)
```

### 2. Root Mean Squared Error



### Actual VS Predicted

```python
y_pred = np.array([1.2, 2.1, 3.5, 4.8])
y_actual = np.array([1, 2, 3, 5])
```

```python
rmse = np.sqrt(((y_pred - y_actual) ** 2).mean())
print('RMSE:', rmse)
```

# Cost Function For Classification

## 1. Binary Classification(Binary Cross-Entropy/Logloss)

****

### Actual VS Predicted

```python
y_pred = np.array([0.2, 0.8, 0.6, 0.3])
y_actual = np.array([0, 1, 1, 0])
n = len(y_pred)
```

```python
bce = -1/n * np.sum(y_actual * np.log(y_pred) + (1 - y_actual) * np.log(1 - y_pred))
print('Binary Cross-Entropy:', bce)
```

### 2. Multiclass Classification(Categorical Cross-Entropy) 
## $$
L_{\text{cross-entropy}}(\hat y,y) = -\sum_{i} y_i \times log(\hat{y}_i)
$$

### Actual VS Predicted

```python
y_pred = np.array([[0.1, 0.3, 0.6], [0.2, 0.7, 0.1], [0.9, 0.05, 0.05]])
y_actual = np.array([[0, 0, 1], [0, 1, 0], [1, 0, 0]])
n = len(y_pred)
```

```python
cce = -1/n * np.sum(np.sum(y_actual * np.log(y_pred)))
print('Categorical Cross-Entropy:', cce)
```
