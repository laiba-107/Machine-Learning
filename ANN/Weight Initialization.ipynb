## Weight Initialization Methods

### Uniform Random Inializer



```python
import numpy as np

# Define the layer sizes
input_size = 5
output_size = 3

# Define the range for uniform initialization
a = -0.1
b = 0.1

# Initialize weights
weights = np.random.uniform(a, b, size=(input_size, output_size))
biases = np.random.uniform(a, b, size=(output_size,))

print("Weights:\n", weights)
print("Biases:\n", biases)
```

### Xavier/Glorot Normal



```python
import numpy as np

def initialize_weights(fan_in, fan_out):
    
    std_dev = np.sqrt(2 / (fan_in + fan_out))
    weights = np.random.normal(loc=0.0, scale=std_dev, size=(fan_in, fan_out))
    return weights

# Example usage:
fan_in = 5
fan_out = 4
W = initialize_weights(fan_in, fan_out)
print(W.shape)
print(W[:5, :5])  # Print first 5x5 block for inspection
```

### Xavier Uniform
!


```python
import numpy as np

def xavier_uniform(fan_in, fan_out):
  
    limit = np.sqrt(6 / (fan_in + fan_out))
    weights = np.random.uniform(low=-limit, high=limit, size=(fan_in, fan_out))
    return weights

# Example usage:
fan_in = 6
fan_out = 64
W = xavier_uniform(fan_in, fan_out)
print(W.shape)
print(W[:5, :5])  # Inspect first 5x5 block
```
