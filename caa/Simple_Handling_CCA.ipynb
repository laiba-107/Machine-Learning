```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# read file
df = pd.read_csv('sample_data.csv')
df
```

```python
df.isnull()
```

```python
df.isnull().mean()
```

```python
cols = [var for var in df.columns if df[var].isnull().mean() < 0.09 and df[var].isnull().mean() > 0 ]
cols
```

```python
dfnew = df.dropna()
dfnew
```

```python
len(df.dropna()) / len(df)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111)

# original data
df['Experience'].hist(bins=50, ax=ax, density=True, color='red')

# data after cca, the argument alpha makes the color transparent, so we can
# see the overlay of the 2 distributions
dfnew['Experience'].hist(bins=50, ax=ax, color='green', density=True, alpha=0.8)
```

```python
df.shape, dfnew.shape
```

```python
# Plot densities
df['Experience'].plot.density(color='red', label='Original')
dfnew['Experience'].plot.density(color='blue', label='After Drop')
plt.legend()
```
