```

## One-Hot Encoding

Convert categorical features into multiple binary columns using `OneHotEncoder`.

```
# Initialize OneHotEncoder
ohe = OneHotEncoder()

# Apply One-Hot Encoding
encoded_array = ohe.fit_transform(df[['COLUMN_NAME']])

# Convert to dense array and create DataFrame
encoded_df = pd.DataFrame(
    encoded_array,
    columns=ohe.get_feature_names_out(['COLUMN_NAME'])
)


## Mean Encoding (Target Encoding)

Encode categorical features based on the **mean of a target variable**.

```
# Compute mean encoding
mean_encoding = df.groupby('COLUMN_NAME')['TARGET_COLUMN'].mean()

# Map mean values to the dataset
df['COLUMN_NAME_Mean_Encoded'] = df['COLUMN_NAME'].map(mean_encoding)

# Display the encoded column
df[['COLUMN_NAME', 'COLUMN_NAME_Mean_Encoded']]

## Frequency / Count Encoding

Encode categorical features based on the **number of occurrences** of each category.

```
# Compute counts for each category
category_counts = df['COLUMN_NAME'].value_counts()

# Map counts to the dataset
df['COLUMN_NAME_Freq'] = df['COLUMN_NAME'].map(category_counts)

# Display the encoded column
df[['COLUMN_NAME', 'COLUMN_NAME_Freq']]
