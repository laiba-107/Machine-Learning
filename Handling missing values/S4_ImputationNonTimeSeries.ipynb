```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data_science_job.csv')
df
```

## Check for null values

```python
df.isnull() #get False for non-null value and True for null value
```

## Median Imputation (Numerical Columns)

More robust to outliers than mean.

```python
median_imputation = df['training_hours'].fillna(df['training_hours'].median())
median_imputation
```

```python
cols = ['city_development_index', 'experience', 'training_hours']
```

```python
median_imputation = df[cols].fillna(df[cols].mean())
median_imputation
```

df['training_hours'] selects the training_hours column as a Series.

.median() computes the median value of that column.

.fillna(...) replaces all missing values (NaN) in that Series with the median.
The result is stored in median_imputation.

```python
df[cols] = median_imputation
df
```

## Mean Imputation (Numerical Columns)

Replaces missing values with the average of the column.

```python
cols = ['city_development_index', 'experience', 'training_hours']
```

```python
mean_imputation = df[cols].fillna(df[cols].mean())
mean_imputation
```

cols is a list of numeric column names, e.g. ['city_development_index', 'experience', 'training_hours'].

df[cols] selects those columns as a DataFrame.

.mean() computes the mean for each selected column, returning a Series of column means.

.fillna(df[cols].mean()) fills missing values in each column using that column’s own mean.

```python
df[cols] = mean_imputation
df
```

## Mode Imputation (Categorical + Numerical)

Replaces missing values with the most frequent value.

```python
Mode_imputation = df['gender'].fillna(df['gender'].mode()[0])
Mode_imputation
```

```python
df['gender'] = Mode_imputation
df
```
