# Simple Neural Network Forward Pass from Scratch

Building a simple neural network using only standard Python (no libraries) based on the provided dataset.

```python
# Dataset from the table (Input 1, Input 2, Output)
data = [
    [1, 4, 1],
    [0.6, 0.7, 0],
    [0.8, 0.9, 0],
    [3, 2, 1],
    [4, 6, 2]
]

# Initial Weights and Biases (from user snippet)
W1 = 0.72
B1 = 0.5
W2 = 0.45
B2 = 0.35

print(f"{'Item':<5} {'X1':<10} {'X2':<10} {'H1':<10} {'OH1':<10} {'H2':<10} {'OH2':<10} {'Actual':<10}")
print("-" * 85)

for i, row in enumerate(data):
    X1, X2, actual = row
    
    # Calculation for First Input (H1)
    H1 = (W1 * X1) + B1
    OH1 = max(H1, 0) # ReLU Activation
    
    # Calculation for Second Input (H2)
    H2 = (W2 * X2) + B2
    OH2 = max(H2, 0) # ReLU Activation
    
    print(f"{i+1:<5} {X1:<10} {X2:<10} {H1:<10.4f} {OH1:<10.4f} {H2:<10.4f} {OH2:<10.4f} {actual:<10}")
```
