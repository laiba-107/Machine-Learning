# Data Preprocessing and Missing Value Imputation

This notebook demonstrates how to load a dataset, identify missing values, apply different imputation techniques, and visualize the results using Python.


## 1. Import Required Libraries

In this step, we import essential Python libraries used for data manipulation, numerical computation, and visualization.

- **NumPy**: For numerical operations  
- **Pandas**: For data handling and analysis  
- **Matplotlib**: For data visualization  

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Read the Sample Dataset

The dataset is loaded from a CSV file named `sample_data.csv` using pandas.

```python
df = pd.read_csv("sample_data.csv")
```

## 3. Check for Missing (Null) Values by using isna() or isnull() functions

We examine the dataset to identify whether any columns contain missing (null) values.

- Display the count of null values for each column  
- Determine which columns require imputation  

```python
df.info()
df
```

```python
df.isnull()
```

## 4. Handle Missing Values in Columns

If a particular column contains missing values, the column will be filled using an appropriate imputation technique.  
If multiple columns contain missing values, all such columns will be handled accordingly.

```python
#df.dropna() #not going to user as almost all columns will be dropped
cols = ['Age', 'Salary', 'Experience']
cols
```

## 5. Median Imputation

Median imputation is applied to **numerical columns**.

- Suitable when data contains outliers  
- Replaces missing values with the median of the column  

```python
#for a single column: Age 
med_imputation = df['Age'].fillna(df['Age'].median())
med_imputation
```

```python
median_imputation = df[cols].fillna(df[cols].median())
median_imputation
```

## 6. Mean Imputation

Mean imputation is also applied to **numerical columns**.

- Best suited for normally distributed data  
- Replaces missing values with the average (mean) of the column  

```python
#df.fillna(df.mean(), inplace=True)
# df['Age'].fillna(df['Age'].mean(), inplace=True)
# df['Experience'].fillna(df['Experience'].mean(), inplace=True)

#Another way
mean_imputation = df[cols].fillna(df[cols].mean()[0])
mean_imputation
```

## 7. Mode Imputation for Categorical Data

Mode imputation is used for **categorical columns**.

- Replaces missing values with the most frequent category  
- Commonly used for nominal and ordinal features  

```python
#df.fillna(df.mode(), inplace=True)
mode_imputation = df['City'].fillna(df['City'].mode())
mode_imputation
```

## 8. Visualization of Results

After applying imputation techniques, visualizations are created to compare:

- Data distribution before and after imputation  
- Impact of mean, median, and mode imputation  

Graphs and plots help in understanding how missing value treatment affects the dataset.


```python
plt.plot(med_imputation)
plt.plot(mean_imputation)
# plt.plot(mode_imputation) # do not have numeric values so cannot plot
plt.show()
```
