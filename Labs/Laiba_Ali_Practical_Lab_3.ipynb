## Lab 3.4: Identifying Viral Content Patterns

```python
#sorting the values in ascending
data.sort_values("Likes", ascending=False)
```

```python
# Viral Post
# Create Virality column by summing views and comments
data['Virality'] = data['Views'] + data['Comments']

# Sort dataframe by Virality in descending order to find top viral posts
data_sorted = data.sort_values('Virality', ascending=False)

# Display top 10 viral posts
print("Top 10 Viral Posts:")
print(data_sorted[['Virality', 'Views', 'Comments']].head(10))

# Show summary statistics
print(f"\nVirality Statistics:")
print(f"Mean Virality: {data['Virality'].mean():.2f}")
print(f"Max Virality: {data['Virality'].max():.2f}")
print(f"Min Virality: {data['Virality'].min():.2f}")
```

```python
data.groupby("Virality")["Post_ID"].mean().plot(kind="bar")
plt.show()
```

```python
# Find the top viral post by Post_ID
top_viral_post = data_sorted.iloc[0]

print("Top Viral Post:")
print(f"Post_ID: {top_viral_post['Post_ID']}")
print(f"Virality Score: {top_viral_post['Virality']:.2f}")
print(f"Views: {top_viral_post['Views']}")
print(f"Comments: {top_viral_post['Comments']}")

# Alternative: Get top N posts
print("\n\nTop 5 Viral Posts by Post_ID:")
print(data_sorted[['Post_ID', 'Virality', 'Views', 'Comments']].head(5))
```

## Lab 3.5: Designing a Media Dashboard
Overview: Media professionals design dashboards.

```python
platform_data = data.groupby("Platform")[["Likes", "Comments", "Shares"]].sum()
platform_data
```

```python
plt.figure(figsize=(15,10))
```

```python
# Total Likes by Platform
plt.subplot(2,3,1)
plt.bar(platform_data.index, platform_data["Likes"])
plt.title("Total Likes by Platform", fontweight="bold")
plt.xlabel("Platform")
plt.ylabel("Likes")
plt.xticks(rotation=60)
```

```python
#Create Comments chart (Chart 2) Total comments by platform
plt.subplot(2,3,2)
plt.bar(platform_data.index, platform_data["Comments"])
plt.title("Total Comments by Platform")
plt.xlabel("Platform")
plt.xticks(rotation=60)
```

```python
#create share chart Total shares by platform
plt.subplot(2,3,3)
plt.bar(platform_data.index, platform_data["Shares"])
plt.title("Total Shares by Platform")
plt.xlabel("Platform")
plt.ylabel("Shares")
plt.xticks(rotation=60)
```

```python
# Create a dashboard with multiple charts
plt.figure(figsize=(15, 10))

# Chart 1: Total Likes by Platform
plt.subplot(2, 3, 1)
plt.bar(platform_data.index, platform_data["Likes"])
plt.title("Total Likes by Platform", fontweight='bold')
plt.xlabel("Platform")
plt.ylabel("Likes")
plt.xticks(rotation=45)

# Chart 2: Total Comments by Platform
plt.subplot(2, 3, 2)
plt.bar(platform_data.index, platform_data["Comments"])
plt.title("Total Comments by Platform", fontweight='bold')
plt.xlabel("Platform")
plt.ylabel("Comments")
plt.xticks(rotation=45)

# Chart 3: Total Shares by Platform
plt.subplot(2, 3, 3)
plt.bar(platform_data.index, platform_data["Shares"])
plt.title("Total Shares by Platform", fontweight='bold')
plt.xlabel("Platform")
plt.ylabel("Shares")
plt.xticks(rotation=45)

# Display dashboard
plt.tight_layout()
plt.show()
```

## Lab 3.6: Making Media Decisions Using Data
Overview: Media professionals use data to make

```python
#Group data by platform
platform_summary = data.groupby("Platform")[["Views", "Likes", "Comments", "Shares"]].sum()
platform_summary
```

```python
# Calculate Total Engagement
# Total Engagement = Likes + Comments + Shares

platform_summary["Total_Engagement"] = (
platform_summary["Likes"] +
platform_summary["Comments"] +
platform_summary["Shares"]
)
platform_summary
```

```python
#Find the platform with highest engagement
platform_summary.sort_values("Total_Engagement", ascending=False)
```

```python
#Visualise platform performance (recommended)
platform_summary["Total_Engagement"].plot(kind="bar", title="Total Engagement by Platform")
plt.xlabel("Platform")
plt.ylabel("Total Engagement")
plt.show()
```
